```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
plt.rcParams["font.weight"] = "bold"
plt.rcParams["axes.labelweight"] = "bold"
font = {'weight' : 'bold', 'size'   : 20}
plt.rc('font', **font)
colors = list(reversed(sns.color_palette("colorblind")))
#colors = sns.color_palette("colorblind")
hatches = ['/', '\\', '|', '-', '+', 'x', 'o', 'O', '.', '*']
from comparative_analysis import ComparativeAnalysis
analysis = ComparativeAnalysis("continuallearningexperiments.csv")
df = analysis.result_df
df = pd.concat(
    [
        #df[df['training_method'] == 'FedSCDepth(Retrain)'], 
        #df[df['training_method'] == 'FedSCDepth'],
        #df[df['training_method'] == 'FedSCDepth(Average)'],
        #df[df['training_method'] == 'FedSCDepth(Average&Retrain)'],
        df[df['training_method'] == 'BOFedSCDepth(Retrain)'], 
        df[df['training_method'] == 'BOFedSCDepth'],
        df[df['training_method'] == 'BOFedSCDepth(Average)'],
        df[df['training_method'] == 'BOFedSCDepth(Average&Retrain)'],
    ]
)
replace_dict = {
    'FedSCDepth(Average)': 'RFSCD',
    'FedSCDepth(Retrain)': 'ERFSCD',
    'FedSCDepth(Average&Retrain)': 'ERRFSCD',
    'FedSCDepth': 'FSCD',
    'BOFedSCDepth(Average)': 'RBOFSCD',
    'BOFedSCDepth(Retrain)': 'ERBOFSCD',
    'BOFedSCDepth(ConstrainedLoss)': 'LBOFSCD',
    'BOFedSCDepth(ConstrainedLossRetrain)': 'LERBOFSCD',
    'BOFedSCDepth(Average&Retrain)': 'ERRBOFSCD',
    'BOFedSCDepth': 'BOFSCD'
}
df = df.replace({'training_method': replace_dict})

df_ddad = df[df['test_dataset'] == 'DDAD']
df_kitti = df[df['test_dataset'] == 'KITTI']
legend_configs = dict(loc='best', prop={'weight':'bold'}, ncol=2, columnspacing=0.1, fontsize=10, title=None)
#legend_configs = dict(loc="lower right", bbox_to_anchor=(1.0175, 0), prop={'weight':'bold'}, ncol=2, columnspacing=0.1, fontsize=10, title=None)
#legend_configs = dict(loc="upper right", bbox_to_anchor=(1.0175, 1), prop={'weight':'bold'}, ncol=2, columnspacing=0.1, fontsize=10, title=None)
```

```python
analysis.result_df = df_ddad
analysis.plot_metrics_differences(
    ['test_abs_rel', 'test_sq_rel', 'test_rmse', 'test_rmse_log'], 'full', 
    save_as_pdf=True, colors=colors, hatches=hatches, legend_configs=False, label="ddad_err"
)
analysis.plot_metrics_differences(
    ['test_a1', 'test_a2', 'test_a3'], 'full', 
    save_as_pdf=True, colors=colors, hatches=hatches, legend_configs=legend_configs, label="ddad_acc"
)
```

```python
analysis.result_df = df_kitti
analysis.plot_metrics_differences(
    ['test_abs_rel', 'test_sq_rel', 'test_rmse', 'test_rmse_log'], 'full', 
    save_as_pdf=True, colors=colors, hatches=hatches, legend_configs=False, label="kitti_err"
)
analysis.plot_metrics_differences(
    ['test_a1', 'test_a2', 'test_a3'], 'full', 
    save_as_pdf=True, colors=colors, hatches=hatches, legend_configs=legend_configs, label="kitti_acc"
)
```
